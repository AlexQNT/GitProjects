```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def simulate_random_walk(n_steps=3, n_paths=1, p_up=0.5, step_up=1, step_down=-1, seed=42):
    rng = np.random.default_rng(seed)
    steps = rng.choice([step_up, step_down], size=(n_paths, n_steps), p=[p_up, 1 - p_up])
    steps_with_start = np.insert(steps, 0, 0, axis=1)
    paths = np.cumsum(steps_with_start, axis=1)
    return paths


def plot_paths(paths):
    plt.figure(figsize=(10, 6))
    for i in range(paths.shape[0]):
        plt.plot(paths[i])
    plt.xlabel("Step")
    plt.ylabel("Position")
    plt.title("Random Walk Simulation")
    plt.show()
```

```python
paths = simulate_random_walk(n_steps=3, n_paths=1, seed=42)
print(paths)
```

```python
paths = simulate_random_walk(n_steps=500, n_paths=1000, seed=42)
plot_paths(paths)
```

```python
paths = simulate_random_walk(n_steps=100, n_paths=10, seed=42)
plot_paths(paths)
```
